```python
import pandas as pd
import numpy as np


df = pd.read_csv("/content/returns_sustainability_dataset.csv")


df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="%Y-%m-%d")


df["Order_Year"]    = df["Order_Date"].dt.year
df["Order_Month"]   = df["Order_Date"].dt.month
df["Order_Month_Name"] = df["Order_Date"].dt.strftime("%b")   # Jan, Feb …
df["Order_Quarter"] = df["Order_Date"].dt.quarter


df["Returned"] = (df["Return_Status"] == "Returned").astype(int)

df["Refund_Rate"] = np.where(
    df["Order_Value"] > 0,
    (df["Return_Cost"] / df["Order_Value"]).round(4),
    0
)

df["Age_Group"] = pd.cut(
    df["User_Age"],
    bins=[0, 30, 45, 60, 100],
    labels=["18–30", "31–45", "46–60", "60+"],
    right=True
)


df["Discount_Tier"] = pd.cut(
    df["Discount_Applied"],
    bins=[-1, 0, 15, 30, 50],
    labels=["No Discount", "Low (1–15%)", "Medium (16–30%)", "High (31–50%)"]
)


df["Late_Return_Flag"] = np.where(
    df["Returned"] == 1,
    np.where(df["Days_to_Return"] > 7, "Late", "On Time"),
    "Not Returned"
)


df["Return_Speed"] = pd.cut(
    df["Days_to_Return"],D
    bins=[-1, 0, 7, 20, 60],
    labels=["No Return", "Quick (1–7d)", "Standard (8–20d)", "Slow (21–60d)"]
)

df["Return_Reason"] = df["Return_Reason"].str.strip().str.title()

df = df.drop_duplicates(subset="Order_ID")


output_path = "returns_cleaned.xlsx"
df.to_excel(output_path, index=False, engine="openpyxl")

print(f"Saved: {output_path}  |  Rows: {len(df)}  |  Columns: {len(df.columns)}")
```
